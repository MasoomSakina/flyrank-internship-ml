## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Just a placeholder to ensure the code cell executes cleanly
reason_ml_wins = "Handles multi-variable non-linear volatility better than rigid thresholds."
```

A fixed rule might look like: if traffic_drop > 15% AND impressions > 1000, then refresh.
ML beats this because fixed rules cannot differentiate between structural decay and normal seasonal volatility. An ensemble learning model can evaluate dozens of complex, non-linear relationships simultaneously—such as historical baseline variance, temporal momentum, and categorical metadata—to identify true anomalies. It adapts to the unique volatility profile of each page rather than applying a rigid, one-size-fits-all threshold.
